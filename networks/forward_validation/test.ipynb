```python
import random

import numpy as np
import pandas as pd
import torch

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

def threshold_signal(
    predictions,
    threshold,
):
    return np.select(
        [
            predictions > threshold,
            predictions < -threshold,
        ],
        [
            1.0,
            -1.0,
        ],
        default=0.0,
    )

def validation_sharpe(
    predictions,
    actual,
    threshold,
    transaction_cost_bpts,
    periods_per_year,
):
    signal = threshold_signal(
        predictions,
        threshold,
    )

    turnover = np.abs(
        np.diff(
            signal,
            prepend=0.0,
        )
    )

    fee_log = np.log1p(
        -transaction_cost_bpts / 10_000
    )

    net_returns = (
        signal * actual
        + turnover * fee_log
    )

    return_std = np.std(
        net_returns,
        ddof=1,
    )

    if (
        not np.isfinite(return_std)
        or return_std == 0
    ):
        return -np.inf

    return (
        np.mean(net_returns)
        / return_std
        * np.sqrt(periods_per_year)
    )

def tuned_walk_forward(
    returns,
    model_factory,
    lookbacks,
    thresholds,
    initial_train_size=800,
    validation_size=100,
    test_size=100,
    epochs=10_000,
    lr=1e-3,
    weight_decay=0.0,
    transaction_cost_bpts=10,
    periods_per_year=252,
    seeds=(42,),
):
    if isinstance(returns, pd.DataFrame):
        returns = returns.squeeze(
            "columns"
        )

    returns = (
        returns
        .dropna()
        .sort_index()
    )

    lookbacks = sorted(
        set(int(x) for x in lookbacks)
    )

    thresholds = sorted(
        set(float(x) for x in thresholds)
    )

    max_lookback = max(lookbacks)

    # Building all lags up to the maximum gives every
    # candidate identical validation/test dates.
    data = FV.make_lagged_data(
        returns,
        max_lookback,
    )

    test_results = []
    tuning_results = []

    train_end = initial_train_size
    fold_number = 0

    while (
        train_end
        + validation_size
        + test_size
        <= len(data)
    ):
        validation_end = (
            train_end + validation_size
        )

        test_end = (
            validation_end + test_size
        )

        train_df = data.iloc[
            :train_end
        ]

        validation_df = data.iloc[
            train_end:validation_end
        ]

        test_df = data.iloc[
            validation_end:test_end
        ]

        candidates = []

        # ==========================================
        # Select hyperparameters on validation data
        # ==========================================

        for lookback in lookbacks:
            features = [
                f"lag_{lag}"
                for lag in range(
                    1,
                    lookback + 1,
                )
            ]

            seed_predictions = []

            for seed in seeds:
                set_seed(
                    seed + fold_number
                )

                model = model_factory(
                    lookback
                )

                prediction = FV.train_one_fold(
                    model=model,
                    train_df=train_df,
                    test_df=validation_df,
                    features=features,
                    epochs=epochs,
                    lr=lr,
                    weight_decay=weight_decay,
                )

                seed_predictions.append(
                    prediction
                )

            validation_prediction = (
                np.mean(
                    seed_predictions,
                    axis=0,
                )
            )

            actual_validation = (
                validation_df[
                    "target"
                ].to_numpy()
            )

            for threshold in thresholds:
                score = validation_sharpe(
                    predictions=validation_prediction,
                    actual=actual_validation,
                    threshold=threshold,
                    transaction_cost_bpts=(
                        transaction_cost_bpts
                    ),
                    periods_per_year=(
                        periods_per_year
                    ),
                )

                candidate = {
                    "fold": fold_number,
                    "lookback": lookback,
                    "threshold": threshold,
                    "validation_sharpe": score,
                }

                candidates.append(
                    candidate
                )

                tuning_results.append(
                    candidate.copy()
                )

        candidate_df = pd.DataFrame(
            candidates
        )

        best_candidate = candidate_df.loc[
            candidate_df[
                "validation_sharpe"
            ].idxmax()
        ]

        best_lookback = int(
            best_candidate["lookback"]
        )

        best_threshold = float(
            best_candidate["threshold"]
        )

        best_validation_sharpe = float(
            best_candidate[
                "validation_sharpe"
            ]
        )

        # ==========================================
        # Refit on training + validation
        # ==========================================

        train_validation_df = data.iloc[
            :validation_end
        ]

        best_features = [
            f"lag_{lag}"
            for lag in range(
                1,
                best_lookback + 1,
            )
        ]

        test_seed_predictions = []

        for seed in seeds:
            set_seed(
                seed + fold_number
            )

            model = model_factory(
                best_lookback
            )

            prediction = FV.train_one_fold(
                model=model,
                train_df=train_validation_df,
                test_df=test_df,
                features=best_features,
                epochs=epochs,
                lr=lr,
                weight_decay=weight_decay,
            )

            test_seed_predictions.append(
                prediction
            )

        test_prediction = np.mean(
            test_seed_predictions,
            axis=0,
        )

        test_signal = threshold_signal(
            test_prediction,
            best_threshold,
        )

        fold_result = pd.DataFrame(
            {
                "fold": fold_number,
                "predictions": (
                    test_prediction
                ),
                "T_test": (
                    test_df[
                        "target"
                    ].to_numpy()
                ),
                "signal": test_signal,
                "selected_lookback": (
                    best_lookback
                ),
                "selected_threshold": (
                    best_threshold
                ),
                "validation_sharpe": (
                    best_validation_sharpe
                ),
            },
            index=test_df.index,
        )

        test_results.append(
            fold_result
        )

        print(
            f"Fold {fold_number}: "
            f"lookback={best_lookback}, "
            f"threshold={best_threshold:.6f}, "
            f"validation Sharpe="
            f"{best_validation_sharpe:.3f}"
        )

        train_end += test_size
        fold_number += 1

    if not test_results:
        raise ValueError(
            "Not enough data to create "
            "a complete walk-forward fold."
        )

    results = (
        pd.concat(test_results)
        .sort_index()
    )

    # Calculate costs continuously across test folds.
    results["turnover"] = (
        results["signal"]
        .diff()
        .abs()
        .fillna(
            results["signal"].abs()
        )
    )

    fee_log = np.log1p(
        -transaction_cost_bpts / 10_000
    )

    results["tx_fees"] = (
        results["turnover"]
        * fee_log
    )

    results["trade_log_return"] = (
        results["signal"]
        * results["T_test"]
        + results["tx_fees"]
    )

    tuning_results = pd.DataFrame(
        tuning_results
    )

    return results, tuning_results
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Use 365 for daily crypto and 252 for daily equities.
periods_per_year = 365

def calculate_metrics(df, periods_per_year=365):
    df = df.sort_index().copy()

    actual = df["T_test"]
    prediction = df["predictions"]
    signal = df["signal"]
    net_log_return = df["trade_log_return"]

    # ----------------------------------------------
    # Prediction metrics
    # ----------------------------------------------

    zero_mse = np.mean(actual**2)
    model_mse = np.mean(
        (actual - prediction) ** 2
    )

    mse_ratio = (
        model_mse / zero_mse
        if zero_mse > 0
        else np.nan
    )

    oos_r2 = 1 - mse_ratio

    pearson_ic = prediction.corr(
        actual,
        method="pearson",
    )

    spearman_ic = prediction.corr(
        actual,
        method="spearman",
    )

    directional_accuracy = (
        np.sign(prediction)
        == np.sign(actual)
    ).mean()

    # ----------------------------------------------
    # Trading metrics
    # ----------------------------------------------

    active = signal != 0

    active_win_rate = (
        (
            signal[active]
            * actual[active]
        ) > 0
    ).mean() if active.any() else np.nan

    active_fraction = active.mean()
    long_fraction = (signal == 1).mean()
    short_fraction = (signal == -1).mean()
    neutral_fraction = (signal == 0).mean()

    gross_log_return = (
        signal * actual
    )

    gross_return = np.expm1(
        gross_log_return.sum()
    )

    net_return = np.expm1(
        net_log_return.sum()
    )

    return_std = net_log_return.std()

    annualized_volatility = (
        return_std
        * np.sqrt(periods_per_year)
    )

    annualized_sharpe = (
        net_log_return.mean()
        / return_std
        * np.sqrt(periods_per_year)
        if (
            np.isfinite(return_std)
            and return_std > 0
        )
        else np.nan
    )

    # Drawdown starts from initial log equity = 0.
    equity_log = net_log_return.cumsum()

    peak_log = (
        equity_log
        .cummax()
        .clip(lower=0)
    )

    drawdown_log = (
        equity_log - peak_log
    )

    max_drawdown = np.expm1(
        drawdown_log.min()
    )

    total_turnover = df["turnover"].sum()

    position_changes = (
        df["turnover"] > 0
    ).sum()

    transaction_cost = np.expm1(
        df["tx_fees"].sum()
    )

    return pd.Series({
        "observations": len(df),

        # Prediction quality
        "pearson_ic": pearson_ic,
        "spearman_ic": spearman_ic,
        "zero_mse": zero_mse,
        "model_mse": model_mse,
        "mse_ratio": mse_ratio,
        "oos_r2": oos_r2,
        "directional_accuracy": (
            directional_accuracy
        ),

        # Positions
        "active_win_rate": active_win_rate,
        "active_fraction": active_fraction,
        "long_fraction": long_fraction,
        "short_fraction": short_fraction,
        "neutral_fraction": neutral_fraction,

        # Trading performance
        "gross_return": gross_return,
        "net_return": net_return,
        "annualized_volatility": (
            annualized_volatility
        ),
        "annualized_sharpe": (
            annualized_sharpe
        ),
        "max_drawdown": max_drawdown,
        "total_turnover": total_turnover,
        "position_changes": position_changes,
        "transaction_cost": transaction_cost,
    })

# ==================================================
# 1. Test metrics for every outer fold
# ==================================================

fold_metrics = pd.DataFrame(
    [
        {
            "fold": fold_number,
            **calculate_metrics(
                fold_data,
                periods_per_year,
            ).to_dict(),
        }
        for fold_number, fold_data
        in results.groupby("fold")
    ]
).set_index("fold")

display(fold_metrics.round(4))

# ==================================================
# 2. Pooled out-of-sample test metrics
# ==================================================

pooled_metrics = calculate_metrics(
    results,
    periods_per_year,
).to_frame("pooled_test")

display(pooled_metrics.round(4))

# ==================================================
# 3. Hyperparameters selected in each fold
# ==================================================

selected_settings = (
    results
    .groupby("fold")
    .agg(
        selected_lookback=(
            "selected_lookback",
            "first",
        ),
        selected_threshold=(
            "selected_threshold",
            "first",
        ),
        validation_sharpe=(
            "validation_sharpe",
            "first",
        ),
    )
)

display(selected_settings)

# Selection frequencies
lookback_frequency = (
    selected_settings[
        "selected_lookback"
    ]
    .value_counts(normalize=True)
    .sort_index()
    .rename("frequency")
)

threshold_frequency = (
    selected_settings[
        "selected_threshold"
    ]
    .value_counts(normalize=True)
    .sort_index()
    .rename("frequency")
)

print("Selected-lookback frequency:")
display(lookback_frequency)

print("Selected-threshold frequency:")
display(threshold_frequency)

# ==================================================
# 4. Validation grid summary
# ==================================================

validation_summary = (
    tuning_results
    .groupby(
        ["lookback", "threshold"]
    )["validation_sharpe"]
    .agg(["mean", "std", "median", "count"])
    .sort_values(
        "mean",
        ascending=False,
    )
)

display(validation_summary.round(4))

# ==================================================
# 5. Equity and drawdown curves
# ==================================================

plot_results = results.sort_index().copy()

plot_results["equity"] = np.exp(
    plot_results[
        "trade_log_return"
    ].cumsum()
)

plot_results["underlying_equity"] = np.exp(
    plot_results["T_test"].cumsum()
)

plot_results["equity_peak"] = (
    plot_results["equity"]
    .cummax()
    .clip(lower=1.0)
)

plot_results["drawdown"] = (
    plot_results["equity"]
    / plot_results["equity_peak"]
    - 1
)

fig, axes = plt.subplots(
    3,
    2,
    figsize=(16, 14),
)

# Equity curve
axes[0, 0].plot(
    plot_results.index,
    plot_results["equity"],
    label="Tuned strategy",
    color="red",
)

axes[0, 0].plot(
    plot_results.index,
    plot_results["underlying_equity"],
    label="Underlying",
    color="black",
)

axes[0, 0].axhline(
    1,
    color="grey",
    linewidth=1,
)

axes[0, 0].set_title(
    "Out-of-sample equity"
)

axes[0, 0].set_ylabel(
    "Equity multiplier"
)

axes[0, 0].legend()
axes[0, 0].grid(alpha=0.3)

# Drawdown
axes[0, 1].fill_between(
    plot_results.index,
    plot_results["drawdown"],
    0,
    color="red",
    alpha=0.3,
)

axes[0, 1].set_title(
    "Out-of-sample drawdown"
)

axes[0, 1].set_ylabel("Drawdown")
axes[0, 1].grid(alpha=0.3)

# IC by fold
axes[1, 0].plot(
    fold_metrics.index,
    fold_metrics["pearson_ic"],
    marker="o",
    label="Pearson IC",
)

axes[1, 0].plot(
    fold_metrics.index,
    fold_metrics["spearman_ic"],
    marker="o",
    label="Spearman IC",
)

axes[1, 0].axhline(
    0,
    color="black",
    linewidth=1,
)

axes[1, 0].set_title("IC by test fold")
axes[1, 0].set_xlabel("Fold")
axes[1, 0].set_ylabel("IC")
axes[1, 0].legend()
axes[1, 0].grid(alpha=0.3)

# Sharpe by fold
axes[1, 1].bar(
    fold_metrics.index,
    fold_metrics["annualized_sharpe"],
    color="steelblue",
)

axes[1, 1].axhline(
    0,
    color="black",
    linewidth=1,
)

axes[1, 1].set_title(
    "Net Sharpe by test fold"
)

axes[1, 1].set_xlabel("Fold")
axes[1, 1].set_ylabel("Annualized Sharpe")
axes[1, 1].grid(alpha=0.3)

# Selected lookback
axes[2, 0].step(
    selected_settings.index,
    selected_settings[
        "selected_lookback"
    ],
    where="mid",
    marker="o",
)

axes[2, 0].set_title(
    "Selected lookback by fold"
)

axes[2, 0].set_xlabel("Fold")
axes[2, 0].set_ylabel("Lookback")
axes[2, 0].grid(alpha=0.3)

# Selected threshold
axes[2, 1].step(
    selected_settings.index,
    selected_settings[
        "selected_threshold"
    ],
    where="mid",
    marker="o",
)

axes[2, 1].set_title(
    "Selected threshold by fold"
)

axes[2, 1].set_xlabel("Fold")
axes[2, 1].set_ylabel(
    "Threshold"
)

axes[2, 1].grid(alpha=0.3)

fig.suptitle(
    "Tuned walk-forward test analysis",
    fontsize=16,
)

fig.tight_layout()
plt.show()
```
